```python
import pandas as pd
import numpy as np
df1 = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika'],
    'Department': ['Marketing', 'IT', 'IT'],
    'Salary': [35000, 80000, 75000]
})

df2 = pd.DataFrame({
    'Employee Name': ['Pari', 'Rishabh', 'Neha'],
    'Department': ['IT', 'HR', 'Finance'],
    'Salary': [55000, 60000, 90000]
})

customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['Raj', 'Ayush', 'Pari', 'Neha']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104],
    'customer_id': [1, 2, 5, 3],
    'amount': [500, 300, 700, 200]
})
```

**Q1.** Concatenate `df1` and `df2` vertically (append rows).

```python
pd.concat([df1,df2])
```

**Q2.** Concatenate `df1` and `df2` vertically and reset the index.

```python
pd.concat([df1,df2]).reset_index()
```

**Q3.** Concatenate `df1` and `df2` horizontally (side by side).

```python
pd.concat([df1,df2], axis=1)
```

**Q4.** Perform an inner join between `customers` and `orders` using `customer_id`.

```python
pd.merge(customers,orders,on='customer_id',how='inner')
```

**Q5.** Perform a left join so that customers who have no orders are also included.

```python
pd.merge(customers,orders,on='customer_id',how='left')
```

**Q6.** Perform a right join between `customers` and `orders`.

```python
pd.merge(customers,orders,on='customer_id',how='right')
```

**Q7.** Perform an outer join between `customers` and `orders` and use `indicator=True`.

```python
pd.merge(orders,customers,on='customer_id',how='outer',indicator=True)
```

**Q8.** Concatenate `df1` and `df2` using `keys=['batch1', 'batch2']`, then extract the `batch1` data using `.loc`.

```python
result = pd.concat([df1,df2], keys=['batch1','batch2'])
result.loc['batch1']
```

**Q9.** Merge two DataFrames where the join column names are different. Use `left_on` and `right_on`.

```python
pd.merge(customers, orders,
         left_on='customer_id',
         right_on='cust_id')
```

**Q10.** After merging `customers` and `orders`, calculate the total order amount for each customer.

```python
total_or_amo = pd.merge(customers,orders)
total_or_amo.groupby('customer_id')['amount'].sum()
```
