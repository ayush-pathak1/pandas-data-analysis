```python
import pandas as pd
import numpy as np
df1 = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika'],
    'Department': ['Marketing', 'IT', 'IT'],
    'Salary': [35000, 80000, 75000]
})

df2 = pd.DataFrame({
    'Employee Name': ['Pari', 'Rishabh', 'Neha'],
    'Department': ['IT', 'HR', 'Finance'],
    'Salary': [55000, 60000, 90000]
})

customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['Raj', 'Ayush', 'Pari', 'Neha']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104],
    'customer_id': [1, 2, 5, 3],
    'amount': [500, 300, 700, 200]
})
```

1 .Concatenate df1 and df2, then filter only the rows where the Department is IT.

```python
result = pd.concat([df1,df2])
result[result['Department'] == 'IT']
```

2. Perform an inner join between customers and orders using customer_id, then filter the rows where amount is greater than 400.

```python
result = pd.merge(customers,orders,on='customer_id',how='inner')
result[ result['amount'] > 400 ]
```

Q3. Perform a left join, then identify which customers do not have any orders (rows where order information is NaN).

```python
result = pd.merge(customers,orders,on='customer_id',how='left')
result[result['order_id'].isna()]
```

Q4. Perform an outer join with indicator=True, then return only the rows where _merge == 'left_only'

```python
result  = pd.merge(customers,orders,on='customer_id',how='outer',indicator=True)
result[result['_merge'] == 'left_only']
```

Q5. Merge customers and orders, then calculate the total order amount for each customer, and finally sort the result in descending order.

```python
result = pd.merge(customers,orders,on='customer_id',how='outer')
result.groupby('customer_id')['amount'].sum().sort_values(ascending=False)
```

Q6. Concatenate df1 and df2 using keys=['team1', 'team2'], then extract the team2 data.

```python
result = pd.concat([df1,df2],keys=['team1', 'team2'])
result.loc['team2']
```

7. Concatenate df1 and df2, then calculate the average salary for each department

```python
result =pd.concat([df1,df2], ignore_index=True)
result.groupby('Department')['Salary'].mean()
```

Q8. Merge customers and orders, then keep only the name column from customers and the amount column from orders in the final result.

```python
result = pd.merge(customers,orders,on='customer_id',how='outer')
result[['name','amount']]
```

Q9. Horizontally concatenate df1 and df2. First predict how many columns the resulting DataFrame will have, then run the code and verify your answer.

```python
pd.concat([df1,df2],axis=1)
```

Q10. Merge customers and orders, calculate the total amount for each customer, and then filter only the customers whose total amount is greater than 400.

```python
result = pd.merge(customers,orders,on='customer_id',how='outer')
total_amt = result.groupby('customer_id')['amount'].sum()
total_amt[total_amt > 400]
```
