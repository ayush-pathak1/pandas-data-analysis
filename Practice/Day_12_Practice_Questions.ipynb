```python
import pandas as pd
import numpy as np
df1 = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika'],
    'Department': ['Marketing', 'IT', 'IT'],
    'Salary': [35000, 80000, 75000]
})

df2 = pd.DataFrame({
    'Employee Name': ['Pari', 'Rishabh', 'Neha'],
    'Department': ['IT', 'HR', 'Finance'],
    'Salary': [55000, 60000, 90000]
})

customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['Raj', 'Ayush', 'Pari', 'Neha']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104],
    'customer_id': [1, 2, 5, 3],
    'amount': [500, 300, 700, 200]
})
```

Q1. Concatenate df1 and df2, then filter the rows where Salary is greater than 60000.

```python
result = pd.concat([df1,df2])
result[result['Salary'] > 60000]
```

Q2. Perform an inner join between customers and orders, then calculate the total sum of amount as a single number.

```python
result = pd.merge(customers,orders,on='customer_id', how='inner')
result['amount'].sum()
```

Q3. Perform a left join between customers and orders, then return the name of customers who do not have any orders.

```python
result = pd.merge(customers,orders,on='customer_id', how='left')
result[result['order_id'].isna()]['name']
```

Q4: Perform an outer join between customers and orders with indicator=True, then return only the rows where _merge == 'right_only'.

```python
result = pd.merge(customers,orders,on='customer_id', how='outer',indicator=True)
result[result['_merge'] == 'right_only']
```

Q5.Merge customers and orders, calculate the average amount for each customer, and sort the result in descending order.

```python
result = pd.merge(customers,orders,on='customer_id', how='outer')
result.groupby('customer_id')['amount'].mean().sort_values(ascending=False)
```

Q6. Concatenate df1 and df2 using keys=['batch1', 'batch2'], then count the rows in batch1.

```python
result = pd.concat([df1,df2],keys=['batch1','batch2'])
result.loc['batch1'].shape[0]
```

Q7. Concatenate df1 and df2, then calculate the maximum salary for each department.

```python
result = pd.concat([df1,df2])
result.groupby('Department')['Salary'].max()
```

Q8.Merge customers and orders, then filter the rows where amount is between 300 and 600 (use &).

```python
result = pd.merge(customers,orders,on='customer_id',how='outer')
result[(result['amount'] >= 300) & (result['amount'] <= 600)]
```

Q9. perform a left join between customers and orders, remove rows with missing order_id, and find how many rows are left.

```python
result = pd.merge(customers,orders,on='customer_id',how='left')
result.dropna(subset=['order_id'],inplace=True)
len(result)
```

Q10.Merge → groupby name → calculate total amount → keep totals greater than 400 → sort the result in ascending order.

```python
result =pd.merge(customers,orders,on='customer_id',how='outer')
total_amt = result.groupby('name')['amount'].sum()
total_amt[total_amt > 400].sort_values(ascending=True)
```
