```python
import pandas as pd
import numpy as np
products = pd.DataFrame({
    'product_id': [1, 2, 3, 4, 5],
    'product_name': ['Laptop', 'Phone', 'Tablet', 'Watch', 'Earbuds'],
    'category': ['Electronics', 'Electronics', 'Electronics', 'Accessories', 'Accessories'],
    'price': [50000, 30000, 25000, 8000, 3000]
})

sales = pd.DataFrame({
    'sale_id': [101, 102, 103, 104, 105, 106],
    'product_id': [1, 2, 1, 6, 3, 2],
    'quantity': [2, 1, 3, 1, 2, 4],
    'city': ['Delhi', 'Mumbai', 'Pune', 'Delhi', 'Mumbai', 'Pune']
})
```

Q1.Perform an inner join between products and sales on product_id to display only the products that have sales.

```python
df = pd.merge(products,sales,on="product_id",how="inner")
df
```

Q2.Perform a left join between products and sales, then find the products that have no sales by filtering rows where sale_id is missing.

```python
df = pd.merge(products,sales,on="product_id",how='left')
df[df['sale_id'].isna()]
```

Q3.Perform a left join and return only the product_name of products that have no sales.

```python
df = pd.merge(products,sales,on="product_id",how='left')
df[df['sale_id'].isna()]['product_name']
```

Q4.Perform an outer join between products and sales using indicator=True, then return only the rows where _merge is right_only.

```python
df = pd.merge(products,sales,on='product_id',how='outer',indicator=True)
df[df['_merge'] == 'right_only']
```

Q5.Perform an inner join between products and sales, then filter rows where price is between 10,000 and 40,000 (inclusive). Use & to combine the conditions

```python
df = pd.merge(products,sales,on='product_id',how='inner')
df[(df['price'] >= 10000) & (df['price'] <= 40000)]
```

Q6.Perform an inner join between products and sales, then calculate the average price for each category.

```python
df = pd.merge(products,sales,on='product_id',how='inner')
df.groupby('category')['price'].mean()
```

Q7.Perform an inner join between products and sales, then calculate the total quantity for each city and sort the result in descending order.

```python
df = pd.merge(products,sales,on='product_id',how='inner')
df.groupby('city')['quantity'].sum().sort_values(ascending=False)
```

Q8.Perform a left join between products and sales, drop rows where sale_id is missing using dropna(), and count the remaining rows.

```python
df = pd.merge(products,sales,on='product_id',how='left')
sold_product = df.dropna(subset=['sale_id'])
len(sold_product)
```

Q9.Perform an inner join between products and sales, group by product_id, calculate the total quantity, filter totals greater than 2, and sort the result in ascending order.

```python
df = pd.merge(products,sales,on='product_id',how='inner')
total_quantity = df.groupby('product_id')['quantity'].sum()
total_quantity[total_quantity > 2].sort_values(ascending=True)
```

Q10.Perform an inner join between products and sales, group by category, calculate the average price, filter averages greater than 15,000, and sort the result in descending order.

```python
df = pd.merge(products,sales,on='product_id',how='inner')
avg_price = df.groupby('category')['price'].mean()
avg_price[avg_price > 15000].sort_values(ascending=False)
```
