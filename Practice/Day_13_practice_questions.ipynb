```python
import pandas as pd
import numpy as np
df1 = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika'],
    'Department': ['Marketing', 'IT', 'IT'],
    'Salary': [35000, 80000, 75000]
})

df2 = pd.DataFrame({
    'Employee Name': ['Pari', 'Rishabh', 'Neha'],
    'Department': ['IT', 'HR', 'Finance'],
    'Salary': [55000, 60000, 90000]
})

customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['Raj', 'Ayush', 'Pari', 'Neha']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104],
    'customer_id': [1, 2, 5, 3],
    'amount': [500, 300, 700, 200]
})
```

Q1. Perform a left join between customers and orders, then return the name of customers who have no orders.

```python
result = pd.merge(customers,orders,on='customer_id',how='left')
result[result['order_id'].isna()]['name']
```

Q2. Merge customers and orders, calculate the average amount for each customer_id, and sort the result in descending order.

```python
result = pd.merge(customers,orders,on='customer_id',how='outer')
result.groupby("customer_id")['amount'].mean().sort_values(ascending=False)
```

Q3.Perform a left join, remove rows where order_id is missing using dropna(), and count the remaining rows.

```python
result = pd.merge(customers,orders,on='customer_id',how='left')
result.dropna(subset=['order_id'],inplace=True)
len(result)
```

Q4.Perform an outer join with indicator=True, then return only the rows where _merge is right_only.

```python
result = pd.merge(customers,orders,on='customer_id',how='outer',indicator=True)
result[result['_merge'] == 'right_only']
```

Q5. Merge customers and orders, calculate the average amount for each customer_id, and keep only customers whose average amount is greater than 300.

```python
result = pd.merge(customers,orders,on='customer_id',how='outer')
avg_amount = result.groupby('customer_id')['amount'].mean()
avg_amount[avg_amount > 300]
```

Q6. Perform a left join and return only the name column for customers whose order_id is missing.

```python
result = pd.merge(customers,orders,on='customer_id',how='left')
result[result['order_id'].isna()]['name']
```

Q10.Merge customers and orders, calculate the average amount for each customer, filter averages greater than 250, and sort them in descending order.

```python
result = pd.merge(customers,orders,on='customer_id',how='outer')
avg_amt = result.groupby('customer_id')['amount'].mean()
avg_amt[avg_amt > 250].sort_values(ascending=False)
```
