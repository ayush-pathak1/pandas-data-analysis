```python
import pandas as pd
import numpy as np
employees = pd.DataFrame({
    'emp_id': [1, 2, 3, 4, 5],
    'name': ['Raj', 'Priya', 'Amit', 'Neha', 'Vikram'],
    'department': ['IT', 'HR', 'IT', 'Finance', 'HR']
})

salaries = pd.DataFrame({
    'emp_id': [1, 2, 3, 6],
    'salary': [75000, 35000, 50000, 90000]
})
```

Q1. T1 — Salary Range Filter
Perform an inner join between employees and salaries on emp_id, then filter employees whose salary is between 40,000 and 80,000, inclusive.

```python
df = pd.merge(employees,salaries,on='emp_id',how='inner')
df[(df['salary'] >= 40000) & (df['salary'] <=80000)]
```

Q2. T2 — Missing Salary
Perform a left join between employees and salaries, then return the name of employees whose salary is missing.

```python
df = pd.merge(employees,salaries,on='emp_id',how='left')
df[df['salary'].isna()]['name']
```
