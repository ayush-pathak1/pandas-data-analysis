```python
import pandas as pd
import numpy as np

q1_sales = pd.DataFrame({
    'order_id': [1, 2, 3],
    'product': ['Laptop', 'Phone', 'Tablet'],
    'revenue': [50000, 30000, 25000],
    'region': ['North', 'South', 'North']
})

q2_sales = pd.DataFrame({
    'order_id': [4, 5, 6],
    'product': ['Watch', 'Earbuds', 'Laptop'],
    'revenue': [8000, 3000, 45000],
    'region': ['East', 'West', 'East']
})

customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 5],
    'name': ['Raj', 'Priya', 'Amit', 'Neha', 'Vikram'],
    'city': ['Delhi', 'Mumbai', 'Delhi', 'Pune', 'Mumbai']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104],
    'customer_id': [1, 2, 5, 7],
    'amount': [1500, 800, 2200, 500]
})
```

Q1/7. Concatenate q1_sales and q2_sales vertically, then filter only the rows where region is North.

```python
q1_q2_sales = pd.concat([q1_sales,q2_sales])
north_region = q1_q2_sales[q1_q2_sales['region'] == 'North']
north_region
```

Q2/7
Concatenate q1_sales and q2_sales using ignore_index=True, then calculate the total revenue for each region.

```python
q1_q2_sales = pd.concat([q1_sales,q2_sales],ignore_index=True)
total_revenue_by_region = q1_q2_sales.groupby('region')['revenue'].sum()
print(total_revenue_by_region)
```

Q3/7
Concatenate q1_sales and q2_sales using keys=['Q1', 'Q2'], then extract the Q2 data and report its row count.

```python
total_sales = pd.concat([q1_sales,q2_sales],keys=['Q1','Q2'])
print(total_sales.loc['Q2'])
len(total_sales.loc['Q2'])
```

Q4/7
Perform a left join between customers and orders on customer_id, then return the name of customers who have no orders.

```python
df = pd.merge(customers,orders,on='customer_id',how='left')
df[df['order_id'].isna()]['name']
```

Q5/7
Perform an inner join between customers and orders, calculate the total amount for each city, and sort the result in descending order.

```python
df = pd.merge(customers,orders,on='customer_id',how='inner')
total_amount = df.groupby('city').sum()
total_amount['amount'].sort_values(ascending=False)
```

Q6/7
Perform an outer join between customers and orders using indicator=True, then return only the rows where _merge is right_only.

```python
df = pd.merge(customers,orders,on='customer_id',how='outer',indicator=True)
right_rows = df[df['_merge'] == 'right_only']
print(right_rows)
```

Q7/7
Perform an inner join between customers and orders, then filter rows where amount is between 1,000 and 2,000, inclusive.

```python
df = pd.merge(customers,orders,on='customer_id',how='inner')
filtered_rows = df[(df['amount'] >= 1000) & (df['amount'] <= 2000)]
print(filtered_rows)
```
