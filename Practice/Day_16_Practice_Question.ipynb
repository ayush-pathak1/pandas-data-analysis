```python
import pandas as pd
import numpy as np

q1_sales = pd.DataFrame({
    'order_id': [1, 2, 3],
    'product': ['Laptop', 'Phone', 'Tablet'],
    'revenue': [50000, 30000, 25000],
    'region': ['North', 'South', 'North']
})

q2_sales = pd.DataFrame({
    'order_id': [4, 5, 6],
    'product': ['Watch', 'Earbuds', 'Laptop'],
    'revenue': [8000, 3000, 45000],
    'region': ['East', 'West', 'East']
})

customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 5],
    'name': ['Raj', 'Priya', 'Amit', 'Neha', 'Vikram'],
    'city': ['Delhi', 'Mumbai', 'Delhi', 'Pune', 'Mumbai']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104],
    'customer_id': [1, 2, 5, 7],
    'amount': [1500, 800, 2200, 500]
})
```

1.Concatenate q1_sales and q2_sales using keys=['Q1', 'Q2'], then extract the Q1 data and calculate its total revenue.

```python
df = pd.concat([q1_sales,q2_sales],keys=['Q1','Q2'])
Q1 = df.loc['Q1']
Q1['revenue'].sum()
```

Q2/7
Concatenate q1_sales and q2_sales using ignore_index=True, calculate total revenue for each product, sort the result in descending order, and display the top 3 products using .head(3).

```python
df = pd.concat([q1_sales,q2_sales],ignore_index=True)
total_revenue = df.groupby('product')['revenue'].sum()
sorted_values = total_revenue.sort_values(ascending=False)
sorted_values.head(3)
```

Q3/7
Concatenate q1_sales and q2_sales, then filter orders where revenue is between 10,000 and 50,000, inclusive.

```python
df = pd.concat([q1_sales,q2_sales])
df[(df['revenue'] >= 10000) & (df['revenue'] <= 50000)]
```

Q4/7
Perform an inner join between customers and orders on customer_id. Calculate the average amount for each city, filter cities where the average amount is greater than 1,000, and sort the result in descending order.

```python
df = pd.merge(customers,orders,on='customer_id',how='inner')
avg_amt = df.groupby('city')['amount'].mean()
avg_amt[avg_amt > 1000].sort_values(ascending=False)
```

Q5/7
Perform a left join between customers and orders, then count the customers whose order_id is missing. Use .isna() and .sum().

```python
df = pd.merge(customers,orders,on='customer_id',how='left')
missing_id = df['order_id'].isna().sum()
missing_id
```

Q6/7
Perform an outer join between customers and orders using indicator=True, filter only the rows where _merge is 'both', and display only the name and amount columns.

```python
df = pd.merge(customers,orders,on='customer_id',how='outer',indicator=True)
df[df['_merge'] == 'both'][['name','amount']]
```

Q7/7
Perform an inner join between customers and orders, filter orders where amount is greater than 500, then count how many such orders each city has.

```python
df = pd.merge(customers,orders,on='customer_id',how='inner')
filtered_orders = df[df['amount'] > 500]
each_city_orders = filtered_orders.groupby('city')['order_id']
each_city_orders.count()
```
