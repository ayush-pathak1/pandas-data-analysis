R1. Concatenate jan_orders and feb_orders with keys=['Jan', 'Feb'], extract the Feb data, and find its total revenue.

M1. Which store generated the highest total revenue across both months? Return just the store name.

M2. Return the names of customers who have never made a purchase.

M3. For each segment, find the average spend, keep only segments whose average is above 2000, and sort highest first.

```python
import pandas as pd

jan_orders = pd.DataFrame({
    'order_id': [1, 2, 3, 4],
    'product': ['Shoes', 'Bag', 'Shoes', 'Cap'],
    'revenue': [2500, 1800, 2700, 600],
    'store': ['Delhi', 'Pune', 'Delhi', 'Mumbai']
})

feb_orders = pd.DataFrame({
    'order_id': [5, 6, 7, 8],
    'product': ['Bag', 'Shoes', 'Cap', 'Bag'],
    'revenue': [1900, 2600, 700, 2000],
    'store': ['Pune', 'Mumbai', 'Delhi', 'Pune']
})

customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 5],
    'name': ['Asha', 'Rohan', 'Meera', 'Kabir', 'Sana'],
    'segment': ['Gold', 'Silver', 'Gold', 'Silver', 'Gold']
})

purchases = pd.DataFrame({
    'purchase_id': [11, 12, 13, 14, 15],
    'customer_id': [1, 2, 2, 4, 9],
    'spend': [4000, 1200, 800, 3000, 500]
})
```

Q1 — Concatenate jan_orders and feb_orders using keys=['Jan', 'Feb'], extract the February data, and calculate its total revenue.

```python
df = pd.concat([jan_orders,feb_orders],keys=['Jan','Feb']).loc['Feb']
df['revenue'].sum()
```

Q2 — Concatenate jan_orders and feb_orders using ignore_index=True, then display the top 2 products by total revenue.

```python
df1 = pd.concat([jan_orders,feb_orders],ignore_index=True)
Top_Products = df1.groupby('product')['revenue'].sum().sort_values(ascending=False).head(2)
Top_Products
```

Q3 — Perform a left join between customers and purchases, then count how many customers have no purchases using .isna().sum().

```python
df3 = pd.merge(customers,purchases,on='customer_id',how='left')
missing_purchase = df3['purchase_id'].isna().sum()
missing_purchase
```

Q4 — Perform an inner join between customers and purchases, keep only purchases where spend is greater than 1,000, and count how many such purchases each segment has.

```python
df4 = pd.merge(customers,purchases,on="customer_id",how='inner')
greater_puchases = df4[df4['spend'] > 1000]
greater_puchases.groupby('segment')['purchase_id'].count()
```

Q5 — Which store generated the highest total revenue across January and February? Return only the store name.

```python
highest_total_revenue = pd.concat([jan_orders,feb_orders]).groupby('store')['revenue'].sum()
highest_total_revenue.idxmax()
```

Q6 — Return the names of customers who have never made a purchase.

```python
data = pd.merge(customers,purchases,on="customer_id",how="left")
customer_name = data[data['purchase_id'].isna()]
customer_name["name"]
```

Q7 — For each customer segment, calculate the average spend, keep only segments whose average spend is greater than 2,000, and sort the result from highest to lowest.

```python
customer_seg = pd.merge(customers,purchases,on="customer_id",how="inner")
average_spend = customer_seg.groupby('segment')['spend'].mean()
average_spend[average_spend > 2000].sort_values(ascending=False)
```
