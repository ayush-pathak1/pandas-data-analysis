Q3. Attach restaurant details to the combined orders, keeping only orders whose restaurant exists in restaurants. Then find the total amount per cuisine, highest first.

Q4. Return the names of restaurants that received no orders in either month.

Q5. How many orders in the combined data belong to a rest_id that is not in restaurants?

Q6. Using only matched orders, find the average order amount for each city. Keep only cities where the average is above 600.

Q7. Among matched orders, keep those with an amount of at least 500. Then count how many such orders each cuisine has.

```python
import pandas as pd

restaurants = pd.DataFrame({
    'rest_id': [1, 2, 3, 4, 5],
    'rest_name': ['Spice Hub', 'Pizza Point', 'Biryani House', 'Cafe Mocha', 'Wrap Zone'],
    'cuisine': ['Indian', 'Italian', 'Indian', 'Cafe', 'Fast Food'],
    'city': ['Delhi', 'Mumbai', 'Delhi', 'Pune', 'Mumbai']
})

orders_oct = pd.DataFrame({
    'order_id': [1, 2, 3, 4, 5],
    'rest_id': [1, 2, 1, 3, 5],
    'amount': [450, 600, 520, 800, 300]
})

orders_nov = pd.DataFrame({
    'order_id': [6, 7, 8, 9, 10],
    'rest_id': [2, 3, 3, 1, 6],
    'amount': [650, 750, 900, 500, 400]
})
```

Q1. Combine the October and November orders and find the rest_id with the highest total amount.

```python
oct_nov_orders = pd.concat([orders_oct,orders_nov]).groupby("rest_id")['amount'].sum()
oct_nov_orders.idxmax()
```

Q2. From the combined October and November orders, calculate the total amount for each rest_id, sort in descending order, and display the top 3 restaurants.

```python
total_amount = oct_nov_orders.sort_values(ascending=False)
total_amount.head(3)
```

Q3. Attach restaurant details to the combined October and November orders, keeping only orders whose restaurant exists in restaurants. Then calculate the total order amount for each cuisine and sort from highest to lowest.

```python
oct_nov_orders = pd.concat([orders_oct,orders_nov])
add_restaurant = pd.merge(oct_nov_orders,restaurants,on='rest_id',how='inner')
total_amount = add_restaurant.groupby('cuisine')['amount'].sum().sort_values(ascending=False)
print(total_amount)
```

Q4. Return the names of restaurants that received no orders in either month.

```python
all_orders = pd.concat([orders_oct, orders_nov])
no_orders = pd.merge(restaurants,all_orders,on='rest_id',how='left')
no_orders  = no_orders[no_orders['order_id'].isna()]
print(no_orders['rest_name'])
```

Q5. How many orders in the combined data belong to a rest_id that is not in restaurants?

```python
all_orders = pd.concat([orders_oct,orders_nov])
result = pd.merge(all_orders,restaurants,on='rest_id',how='left')
result[result['rest_name'].isna()]
result['rest_name'].isna().sum()
```

Q6. Using only matched orders, calculate the average order amount for each city. Keep only cities where the average amount is greater than 600.

```python
all_orders = pd.concat([orders_oct, orders_nov])
matched_orders = pd.merge(
    all_orders,
    restaurants,
    on='rest_id',
    how='inner'
)
average_orders = matched_orders.groupby('city')['amount'].mean()
average_orders[average_orders > 600]
```

Q7. Among matched orders, keep those with an amount of at least 500. Then count how many such orders each cuisine has.

```python
all_orders = pd.concat([orders_oct,orders_nov])
matched_orders = pd.merge(all_orders,restaurants,on='rest_id',how='inner')
matched_orders = matched_orders[matched_orders['amount']>= 500].groupby('cuisine').count()
matched_orders['order_id']
```
