```python
import pandas as pd

restaurants = pd.DataFrame({
    'rest_id': [1, 2, 3, 4, 5],
    'rest_name': ['Spice Hub', 'Pizza Point', 'Biryani House', 'Cafe Mocha', 'Wrap Zone'],
    'cuisine': ['Indian', 'Italian', 'Indian', 'Cafe', 'Fast Food'],
    'city': ['Delhi', 'Mumbai', 'Delhi', 'Pune', 'Mumbai']
})

orders_oct = pd.DataFrame({
    'order_id': [1, 2, 3, 4, 5],
    'rest_id': [1, 2, 1, 3, 5],
    'amount': [450, 600, 520, 800, 300]
})

orders_nov = pd.DataFrame({
    'order_id': [6, 7, 8, 9, 10],
    'rest_id': [2, 3, 3, 1, 6],
    'amount': [650, 750, 900, 500, 400]
})
```

Q1 – Level 1: Filtering
Question: Show only the orders where amount is more than 600.
WHO: Not applicable (no grouping).
WHAT: Complete order rows.
HOW: Apply a condition on amount.

```python
all_orders = pd.concat([orders_oct, orders_nov],ignore_index=True)
all_orders[all_orders['amount'] > 600]
```

Q2 – Level 1: Aggregation
Question: Find the total of all order amounts. Return one single number.
- WHO: No grouping.
- WHAT: amount
- HOW: Sum

```python
total_revenue = all_orders['amount'].sum()
total_revenue
```

Q3 – Level 2: GroupBy + Sum
Question: Find the total amount for each rest_id.
WHO: rest_id
WHAT: amount
HOW: sum()

```python
all_orders.groupby('rest_id')['amount'].sum()
```

Q4 – Level 2: GroupBy + Count
Question: Find how many orders each rest_id has.
- WHO: rest_id
- WHAT: order_id
- HOW: Count
Business scenario: The company wants to know how many orders each restaurant received.

```python
restaurants_orders = all_orders.groupby('rest_id')['order_id'].count()
print(restaurants_orders )
```

Q5 – Level 3: GroupBy + Sum + Sorting
Question: Find the total amount for each rest_id, sorted from highest to lowest.
- WHO: rest_id
- WHAT: amount
- HOW: Sum, then sort descending.
Business scenario: The company wants a ranking of restaurants based on their total revenue.

```python
all_orders = pd.concat([orders_oct, orders_nov],ignore_index=True)
revenue_by_restaurant = all_orders.groupby('rest_id')['amount'].sum().sort_values(ascending=False)
print(revenue_by_restaurant)
```

Q6 – Level 3: GroupBy + Sum + Maximum Index
Question: Which rest_id has the highest total amount? Return only the restaurant ID, not the amount.
- WHO: rest_id
- WHAT: amount
- HOW: First calculate sum, then identify the index associated with the highest value.
Business scenario: The company wants to identify its highest-revenue restaurant.

```python
all_orders = pd.concat([orders_oct, orders_nov],ignore_index=True)
revenue_by_restaurant = all_orders.groupby('rest_id')['amount'].sum().idxmax()
print(revenue_by_restaurant)
```

Q7 – Level 3: Inner Join + GroupBy + Sum
Question: Join all_orders with restaurants, keeping only orders whose restaurant exists in restaurants. Then find the total amount for each cuisine.
WHO: cuisine
WHAT: amount
HOW: Inner join → Sum
Business scenario: The company wants to compare total revenue generated by each restaurant cuisine category.

```python
result = pd.merge(all_orders,restaurants,on="rest_id",how="inner")
cuisine_revenue = result.groupby('cuisine')['amount'].sum()
print(cuisine_revenue)
```
