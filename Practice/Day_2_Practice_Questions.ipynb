```python
import pandas as pd
data = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika', 'Pari', 'Rishabh',
                      'Neha', 'Vikram', 'Sneha', 'Arjun', 'Priya'],
    'Age': [24, 23, 20, 20, 19, 28, 32, 25, 27, 22],
    'Department': ['Marketing', 'IT', 'IT', 'IT', 'HR', 
                   'Finance', 'Marketing', 'HR', 'IT', 'Finance'],
    'Salary': [35000, 80000, 75000, 55000, 60000, 
               90000, 72000, 48000, 85000, 67000],
    'City': ['Delhi', 'Mumbai', 'Pune', 'Delhi', 'Bangalore',
             'Mumbai', 'Delhi', 'Pune', 'Bangalore', 'Mumbai'],
    'Experience': [2, 1, 0, 0, 1, 5, 8, 3, 4, 1]
})
data
```

Q1. Find the maximum, minimum, mean, and standard deviation of the Salary column.

```python
data['Salary'].agg(['max','min','mean','std'])
```

Q2. Calculate descriptive statistics for all numeric columns at once using describe().

```python
data[['Age','Salary','Experience']].describe()
```

Q3. Sort the DataFrame by Salary in ascending order.

```python
data.sort_values(by='Salary', ascending=True)
```

Q4. Sort the same DataFrame by Salary in descending order.

```python
data.sort_values(by='Salary', ascending=False)
```

Q5. Filter employees with a salary greater than 50,000.

```python
data[data['Salary'] > 50000]
```

Q6. Filter employees whose age is between 20 and 24 (use the & operator).

```python
data[(data["Age"] >= 20) & (data["Age"] <=24)]
```

Q7. Filter employees working in the IT department.

```python
data[data["Department"]== 'IT']
```

Q8. Filter employees who belong to the IT department OR have a salary greater than 60,000 (use the | operator).

```python
data[(data["Department"]== 'IT') | (data["Salary"] >= 60000)]
```

Q9. Select the first 3 rows and only the Name and Salary columns using .loc.

```python
data = data.reset_index(drop=True)
data.loc[0:2,['Employee Name','Salary']]
```

Q10. Select the last 2 rows using .iloc.

```python
data.iloc[-2:]
```
