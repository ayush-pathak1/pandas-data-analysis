```python
import pandas as pd
data = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika', 'Pari', 'Rishabh', 
                      'Neha', 'Vikram', 'Sneha', 'Arjun', 'Priya'],
    'Age': [24, 23, 20, 20, 19, 28, 32, 25, 27, 22],
    'Department': ['Marketing', 'IT', 'IT', 'IT', 'HR', 
                   'Finance', 'Marketing', 'HR', 'IT', 'Finance'],
    'Salary': [35000, 80000, 75000, 55000, 60000, 
               90000, 72000, 48000, 85000, 67000],
    'City': ['Delhi', 'Mumbai', 'Pune', 'Delhi', 'Bangalore',
             'Mumbai', 'Delhi', 'Pune', 'Bangalore', 'Mumbai'],
    'Experience': [2, 1, 0, 0, 1, 5, 8, 3, 4, 1]
})
data
```

Q1. Select rows 3 to 6 and only the Employee Name, Department, and Salary columns using .loc.

```python
data.loc[3:6,['Employee Name','Department',"Salary"]]
```

Q2. Select the first 4 rows and the first 3 columns using .iloc.

```python

data.iloc[0:4,0:3]
```

Q3. Set the Experience column as the DataFrame index, and then reset it back to default.

```python
data = data.set_index('Experience')
data
```

```python
data = data.reset_index(drop=True)
data
```

Q4. Sort the DataFrame by its index using sort_index().

```python
data.sort_index()
```

Q5. Extract rows where the City column contains the word 'Mumbai' using str.contains().

```python
data[data['City'].str.contains('Mumbai', case=False)]
```
