Q7. Forward fill the missing values in the "Department" column.

```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika', 'Pari', 'Rishabh', 
                      'Neha', 'Vikram', 'Sneha', 'Arjun', 'Priya'],
    'Age': [24, np.nan, 20, 20, 19, 28, np.nan, 25, 27, 22],
    'Department': ['Marketing', 'IT', None, 'IT', 'HR', 
                   'Finance', 'Marketing', None, 'IT', 'Finance'],
    'Salary': [35000, 80000, np.nan, 55000, 60000, 
               90000, 72000, 48000, np.nan, 67000],
    'City': ['Delhi', 'Mumbai', 'Pune', None, 'Bangalore',
             'Mumbai', None, 'Pune', 'Bangalore', 'Mumbai'],
    'Experience': [2, 1, 0, np.nan, 1, 5, 8, 3, np.nan, 1]
})
data
```

Q1. How many missing values are in each column?

```python
data.isnull().sum()
```

Q2. Find the percentage of missing values in each column.

```python
data.isnull().mean() * 100
```

Q3. Drop all rows that contain missing values

```python
data.dropna()
```

Q4. Drop all columns that contain missing values.

```python
data.dropna(axis=1)
```

Q5. Fill the missing values in the "Salary" column with the mean of that column.

```python
data['Salary'].fillna(data['Salary'].mean())
```

Q6. Fill the missing values in the "Age" column with the median.

```python
data['Age'].fillna(data['Age'].median())
```

Q7. Forward fill the missing values in the "Department" column.

```python
data['Department'].ffill()
```
