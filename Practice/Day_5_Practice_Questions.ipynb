```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika', 'Pari', 'Rishabh', 
                      'Neha', 'Vikram', 'Sneha', 'Arjun', 'Priya'],
    'Age': [24, np.nan, 20, 20, 19, 28, np.nan, 25, 27, 22],
    'Department': ['Marketing', 'IT', None, 'IT', 'HR', 
                   'Finance', 'Marketing', None, 'IT', 'Finance'],
    'Salary': [35000, 80000, np.nan, 55000, 60000, 
               90000, 72000, 48000, np.nan, 67000],
    'City': ['Delhi', 'Mumbai', 'Pune', None, 'Bangalore',
             'Mumbai', None, 'Pune', 'Bangalore', 'Mumbai'],
    'Experience': [2, 1, 0, np.nan, 1, 5, 8, 3, np.nan, 1]
})
data
```

Q1. Backward fill the missing values in the "Department" column.

```python
data['Department'].bfill()
```

Q2. Replace the missing values in the "Salary" column with -1 (placeholder value).

```python
data['Salary'].fillna(-1)
```

Q3. Interpolate the missing values in the "Experience" column.

```python
data['Experience'].interpolate()
```

Q4. Replace -99999 with np.nan — first manually insert a few -99999 values into the DataFrame, then perform the replacement.

```python
data = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika', 'Pari', 'Rishabh', 
                      'Neha', 'Vikram', 'Sneha', 'Arjun', 'Priya'],
    'Age': [24, np.nan, 20, -99999, 19, 28, np.nan, 25, -99999, 22],
    'Department': ['Marketing', 'IT', None, 'IT', '-99999', 
                   'Finance', '-99999', None, 'IT', 'Finance'],
    'Salary': [-99999, 80000, np.nan, 55000, 60000, 
               90000, -99999, 48000, np.nan, 67000],
    'City': ['Delhi', 'Mumbai', 'Pune', None, 'Bangalore',
             'Mumbai', None, 'Pune', 'Bangalore', 'Mumbai'],
    'Experience': [-99999, 1, 0, np.nan, 1, 5, 8, 3, np.nan, 1]
})
```

```python
data.replace(-99999,np.nan)
```

Q5. Compare the row counts before and after dropping missing values.

```python
len(data)
```

```python
len(data.dropna())
```

Q6. Keep only those rows that have at least 4 non-null values (dropna(thresh=4)).

```python
data.dropna(thresh=4)
```

Q7. Fill the missing values for both the "City" and "Age" columns simultaneously in a single fillna() call using a dictionary.

```python
data.fillna({
    'City' : 'Unknown',
    'Age' : data['Age'].median()
})
```
