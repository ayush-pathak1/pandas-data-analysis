Q1. First, manually insert $2\text{--}3$ values of -99999 into the DataFrame, then replace them with np.nan.

```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika', 'Pari', 'Rishabh',
                      'Neha', 'Vikram', 'Sneha', 'Arjun', 'Priya'],
    'Age': [24, np.nan, 20, 20, 19, 28, np.nan, 25, 27, 22],
    'Department': ['Marketing', 'IT', None, 'IT', 'HR',
                   'Finance', 'Marketing', None, 'IT', 'Finance'],
    'Salary': [-99999, 80000, np.nan, 55000, -99999,
               90000, 72000, 48000, np.nan, 67000],
    'City': ['  Delhi  ', 'Mumbai', 'Pune', None, 'Bangalore',
             'Mumbai', None, '  Pune  ', 'Bangalore', 'Mumbai'],
    'Experience': [2, 1, 0, np.nan, 1, 5, 8, 3, np.nan, 1]
})
data
```

```python
data = data.replace(-99999,np.nan)
data
```

Q2. Convert the values in the Employee Name column to lowercase using str.lower()

```python
data['Employee Name'] = data['Employee Name'].str.lower()
data
```

Q3. Remove leading and trailing spaces from the City column using str.strip(). (First, manually add extra spaces to a few values).

```python
data['City'] = data['City'].str.strip()
data
```

Q4. Find duplicate rows using duplicated().

```python
data.duplicated()
```

Q5. Remove duplicate rows using drop_duplicates().

```python
data.drop_duplicates()
```

Q6. Convert all values in the Department column to uppercase using str.upper().

```python
data['Department'] = data['Department'].str.upper()
data
```

Q7. Convert the Salary column from string to numeric — first manually convert a few values to strings using astype(str), and then convert them back to numeric using pd.to_numeric()

```python
data['Salary'] = data['Salary'].astype(str)
```

```python
data['Salary'] = pd.to_numeric(data['Salary'],errors='coerce')
data
```
