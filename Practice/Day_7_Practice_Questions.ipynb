```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika', 'Pari', 'Rishabh',
                      'Neha', 'Vikram', 'Sneha', 'Arjun', 'Priya'],
    'Age': [24, 23, 20, 20, 19, 28, 32, 25, 27, 22],
    'Department': ['Marketing', 'IT', 'IT', 'IT', 'HR',
                   'Finance', 'Marketing', 'HR', 'IT', 'Finance'],
    'Salary': [35000, 80000, 75000, 55000, 60000,
               90000, 72000, 48000, 85000, 67000],
    'City': ['Delhi', 'Mumbai', 'Pune', 'Delhi', 'Bangalore',
             'Mumbai', 'Delhi', 'Pune', 'Bangalore', 'Mumbai'],
    'Experience': [2, 1, 0, 0, 1, 5, 8, 3, 4, 1]
})
```

Q1. Rename three columns in a single call — Employee Name → Name, Salary → Monthly_Salary, Experience → Exp_Years.

```python
data.rename(columns={
    'Employee Name' : 'Name',
    'Salary' : 'Monthly_Salary',
    'Experience' : 'Exp_Years'
}, inplace=True)
data
```

Q2. Convert all column names to lowercase.

```python
data.columns = data.columns.str.lower()
data
```

Q3. Replace 'IT' with 'Information Technology' in the Department column.

```python
data['department'] = data['department'].replace('IT','Information Technology')
data
```

Q4. Replace 'Delhi' with 'New Delhi' and 'Pune' with 'Pune City' in the City column — both in a single replace() call.

```python
data['city'].replace({
    'Delhi': 'New Delhi',
    'Pune': 'Pune City'
}, inplace=True)
```

Q5. Create a clean copy of the DataFrame without changing the original (copy()).

```python
copied_data = data.copy()
copied_data
```

Q6. Replace values in the Salary column: values below 50000 with 'Low', 50000–75000 with 'Medium', and values above 75000 with 'High' (use pd.cut()).

```python
data['monthly_salary'] = pd.cut(
    data['monthly_salary'],
    bins=[0,50000,75000,np.inf],
    labels=['Low','Medium','High'],
)
data
```

Q7. Check the dtype of the Age column, then convert it to float.

```python
data['age'].dtype
```

```python
data['age'] = data['age'].astype(float)
data['age'].dtype
```
