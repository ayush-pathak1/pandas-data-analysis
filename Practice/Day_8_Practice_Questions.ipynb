Q1. Calculate the average salary for each department.

Q2. Find the maximum salary for each department.

Q3. Find the minimum salary for each department.

Q4. Count the number of employees in each department.

Q5. Calculate the average age for each city.

Q6. Group by Department and City and calculate the total salary.

Q7. Use agg() in a single call to calculate the mean, minimum, maximum, and count of Salary for each department.

Q8. Find the department with the highest average salary. Use groupby() + idxmax().

Q9. Find all departments where the total salary exceeds 150,000.

Q10. Group by Department and City and calculate the mean age and mean salary together using agg().

```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika', 'Pari', 'Rishabh',
                      'Neha', 'Vikram', 'Sneha', 'Arjun', 'Priya'],
    'Age': [24, 23, 20, 20, 19, 28, 32, 25, 27, 22],
    'Department': ['Marketing', 'IT', 'IT', 'IT', 'HR',
                   'Finance', 'Marketing', 'HR', 'IT', 'Finance'],
    'Salary': [35000, 80000, 75000, 55000, 60000,
               90000, 72000, 48000, 85000, 67000],
    'City': ['Delhi', 'Mumbai', 'Pune', 'Delhi', 'Bangalore',
             'Mumbai', 'Delhi', 'Pune', 'Bangalore', 'Mumbai'],
    'Experience': [2, 1, 0, 0, 1, 5, 8, 3, 4, 1]
})
```

Q1. Calculate the average salary for each department.

```python
data.groupby('Department')['Salary'].mean()
```

Q2. Find the maximum salary for each department.

```python
data.groupby('Department')['Salary'].max()
```

Q3. Find the minimum salary for each department.

```python
data.groupby('Department')['Salary'].min()
```

Q4. Count the number of employees in each department.

```python
data.groupby('Department')['Employee Name'].count()
```

Q5. Calculate the average age for each city.

```python
data.groupby('City')['Age'].mean()
```

Q6. Group by Department and City and calculate the total salary.

```python
data.groupby(['Department','City'])['Salary'].sum()
```

Q7. Use agg() in a single call to calculate the mean, minimum, maximum, and count of Salary for each department.

```python
data.groupby('Department')['Salary'].agg(['mean', 'min', 'max', 'count'])
```

Q8. Find the department with the highest average salary (use groupby + idxmax()).

```python
data.groupby('Department')['Salary'].mean().idxmax()
```

Q9. Find all departments where total salary exceeds 150,000.

```python
result = data.groupby('Department')['Salary'].sum()
result[result > 150000]
```

Q10. Group by two columns (Department and City) and calculate mean age and mean salary together using agg().

```python
data.groupby(['Department','City'])[['Age','Salary']].agg('mean')
```
