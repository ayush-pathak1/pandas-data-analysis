Q10. Group by City and find what percentage of total salary each city contributes (use groupby + sum() + divide by total).

```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'Employee Name': ['Raj', 'Ayush', 'Anamika', 'Pari', 'Rishabh',
                      'Neha', 'Vikram', 'Sneha', 'Arjun', 'Priya'],
    'Age': [24, 23, 20, 20, 19, 28, 32, 25, 27, 22],
    'Department': ['Marketing', 'IT', 'IT', 'IT', 'HR',
                   'Finance', 'Marketing', 'HR', 'IT', 'Finance'],
    'Salary': [35000, 80000, 75000, 55000, 60000,
               90000, 72000, 48000, 85000, 67000],
    'City': ['Delhi', 'Mumbai', 'Pune', 'Delhi', 'Bangalore',
             'Mumbai', 'Delhi', 'Pune', 'Bangalore', 'Mumbai'],
    'Experience': [2, 1, 0, 0, 1, 5, 8, 3, 4, 1]
})
```

Q1. Find total salary by city.

```python
data.groupby('City')['Salary'].agg('sum')
```

Q2. Find average experience by department.

```python
data.groupby('Department')['Experience'].mean()
```

Q3. Count employees in each city.

```python
data.groupby('City')['Employee Name'].count()
```

Q4. Find the employee with the highest salary in each department (use groupby + idxmax()).

```python
data.loc[data.groupby('Department')['Salary'].idxmax()]
```

Q5. Find departments where average age is greater than 22.

```python
DepartmentsAgeIsGreater = data.groupby('Department')['Age'].mean()
DepartmentsAgeIsGreater[DepartmentsAgeIsGreater > 22]
```

Q6. Find the city with the most employees.


```python
data.groupby('City')['Employee Name'].count().idxmax()
```

Q7. Calculate multiple aggregations — for each department find mean salary, max age, and count of employees using agg() with a dictionary.

```python
data.groupby('Department').agg({
    'Salary' : 'mean',
    'Age' : 'max',
    'Employee Name' : 'count'
})
```

Q8. Find total salary and average experience grouped by both Department and City.

```python
data.groupby(['Department','City']).agg({
    'Salary' : 'sum',
    'Experience' : 'mean'
})
```

Q9. Find the department that has the lowest minimum salary.

```python
data.groupby('Department')['Salary'].min().idxmin()
```

Q10.Group by City and find what percentage of total salary each city contributes.

```python
sallary_contro  = data.groupby('City')['Salary'].sum()
sallary_contro / sallary_contro.sum() * 100
```
