```python
import pandas as pd
from pandas.tseries.holiday import USFederalHolidayCalendar
from pandas.tseries.offsets import CustomBusinessDay
df = pd.read_csv("../datasets/aapl_no_dates.csv")
usb = CustomBusinessDay(calendar=USFederalHolidayCalendar())
rng = pd.date_range(start="07/01/2017",end="07/21/2017",freq=usb)
rng
```

```python
df.set_index(rng,inplace=True)
df
```

```python
from pandas.tseries.holiday import AbstractHolidayCalendar, nearest_workday, Holiday
class myCalendar(AbstractHolidayCalendar):
    rules = [
        Holiday('My Birth Day', month=11, day=27),#, observance=nearest_workday),
    ]
    
my_bday = CustomBusinessDay(calendar=myCalendar())
pd.date_range('11/1/2017','11/30/2017',freq=my_bday)
```

```python
b = CustomBusinessDay(weekmask= 'Mon Tue Wed Thu Sun')
rng2 = pd.date_range(start="7/1/2017",end="7/31/2017",freq=b)
rng2
```

```python
b = CustomBusinessDay(weekmask= 'Mon Tue Wed Thu Sun',holidays=["2017-07-04"])
```

```python
rng2 = pd.date_range(start="7/1/2017",end="7/31/2017",freq=b)
rng2
```
