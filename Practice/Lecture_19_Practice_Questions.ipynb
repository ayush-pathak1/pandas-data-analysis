```python
import pandas as pd
dates = ['2023-01-15', '15 Jan 2023', '15/01/2023', '2023.01.15']
pd.to_datetime(dates, format='mixed')
```

```python
dt_us = pd.to_datetime('5/3/2023')
dt_uk = pd.to_datetime('5/3/2023', dayfirst=True)
```

```python
pd.to_datetime('15#03#2023',format='%d#%m#%Y')
```

```python
t = 1609459200
pd.to_datetime(t,unit='s')
```

```python
date = ['2023-01-15', 'not a date', '2023-03-20']
pd.to_datetime(date, errors='coerce')
```

```python
pd.to_datetime('25@12@2023', format='%d@%m@%Y')
```

```python
timestamps = [1609459200, 1640995200, 1672531200]
pd.to_datetime(timestamps,unit='s')
```
