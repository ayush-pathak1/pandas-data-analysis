Create a new column named 'Price Change' calculated as the current day's price minus the previous day's price.

Q5: Calculating 3-Day Return Percentage
Create a new column named '3 day return %' that calculates the 3-day percentage return using shift(3).

```python
import pandas as pd
df = pd.read_csv("../datasets/fb.csv",parse_dates=['Date'],index_col='Date')
df.shift(1)
```

```python
df['Price'].shift(-1)
```

```python
df['Prev Day Price'] = df['Price'].shift(1)
df
```

```python
df['Price Change'] = df['Price'] - df['Prev Day Price']
df
```

```python
df["3 day return %"] = (df['Price'] - df['Price'].shift(3))*100/df['Price'].shift(3)
df
```
