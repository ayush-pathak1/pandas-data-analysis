```python
import pandas as pd
student_df = pd.DataFrame({
    "roll_no" : [1,2,3],
    "name" : ["Ritik","Pari","juju"]
})

marks_df = pd.DataFrame({
    "roll_no" : [1,2,4],
    "marks" : [83,91,76]
})

df = pd.merge(student_df,marks_df, on='roll_no')
df
```

```python
df1 = pd.merge(student_df,marks_df, on='roll_no',how='outer')
df1
```

Question 5 (Combined — merge + pivot_table):
Question 1 wale students_df aur marks_df ko merge karo (inner join), phir result pe pivot_table() lagाओ jisme name index ho aur marks ka simple table dikhे (aggfunc mean se, chahe ek hi value ho).

```python
result = df.pivot_table(index='name', values='marks', aggfunc='mean')
print(result)
```
