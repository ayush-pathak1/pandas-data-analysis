Question 1: Create a basic crosstab with Gender in rows and Product in columns to show counts.

Question 2: Add margins=True to the crosstab from Question 1 to display the Grand Total.

Question 3: Use normalize='index' to find the percentage of Males who buy Product A versus Product B.

Question 4: Use values=data.Age and aggfunc='mean' to calculate the average Age for each Gender-Product combination.

```python
import pandas as pd
data = pd.DataFrame({
    'Gender': ['Male','Female','Male','Female','Male','Female','Male','Female'],
    'Product': ['A','B','A','A','B','B','A','B'],
    'Age': [25,30,22,28,35,26,40,33]
})
data

pd.crosstab(data.Gender,data.Product)
```

```python
pd.crosstab(data.Gender,data.Product,margins=True)
```

```python
pd.crosstab(data.Gender,data.Product,normalize='index')
```

```python
import numpy as np
pd.crosstab([data.Gender],[data.Product],values=data.Age,aggfunc=np.average)
```
