```python
import pandas as pd
from sqlalchemy import create_engine, text
weather_df = pd.read_csv("../datasets/weather_data.csv")
weather_df.to_sql("weather_data", con=engine, if_exists="replace", index=False)

engine = create_engine("sqlite:///../datasets/customer_db.db")
```

```python
query = text("SELECT * FROM weather_data WHERE Temperature > 30")

with engine.connect() as conn:
    df_high_temp = pd.read_sql_query(query, con=conn)

df_high_temp
```

```python
df_weather = pd.read_sql_table("weather_data", con=engine)
df_weather.rename(columns={"Temperature": "Temp_F"}, inplace=True)
df_weather
```

```python
students_df = pd.DataFrame(
    {"name": ["Aarav", "Ayush", "Pari"], "marks": [88, 95, 82]}
)

students_df.to_sql("students", con=engine, if_exists="replace", index=False)
```

```python
df_students = pd.read_sql("students", con=engine)
df_students
```

```python
query_rain = text("SELECT * FROM weather_data WHERE Event = 'Rain'")

with engine.connect() as conn:
    df_rainy_days = pd.read_sql(query_rain, con=conn)

df_rainy_days
```
