Question 5 (resample + Plot): Calculate the monthly average of the Close column and visualize it using a line chart (using .plot(), without kind='bar').

```python
import pandas as pd
df = pd.read_csv("../datasets/aapl.csv", parse_dates=['Date'], index_col=['Date'])
```

```python
df.loc["2016"]
```

```python
df.sort_index()
new_df = df.sort_index()
new_df
```

```python
new_df.Close.resample('W').mean()
```

```python
%matplotlib inline
new_df.Close.resample('ME').mean().plot()
```

```python
new_df.loc["2017-03"].Volume.mean()
```

```python
new_df.Volume.resample('QE').sum()
```
