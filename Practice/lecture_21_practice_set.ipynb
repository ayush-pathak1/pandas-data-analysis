```python
import pandas as pd
df = pd.read_csv("../datasets/msft.csv", header=1, index_col='Date Time', parse_dates=True)
df = df.index.tz_localize(tz='US/Eastern')
df
```

```python
df = df.tz_convert(tz='Europe/Berlin')
df
```

```python
df = df.tz_convert(tz='Asia/Calcutta')
df
```

```python
rng = pd.date_range(start='2017-01-01', periods=10,freq='h',tz='Europe/London')
rng
```

```python
rng = pd.date_range(start='2026-01-01',periods=10,freq='h',tz='dateutil/Europe/London')
rng
```

```python
rng = pd.date_range(start='2026-08-22 09:00:00',periods=10,freq='30min')
s = pd.Series(range(10),index=rng)
s
```

```python
b = s.tz_localize(tz='Europe/Berlin')
b
```

```python
m = s.tz_localize(tz="Asia/Calcutta")
m
```

```python
rng = pd.date_range(
    start='2026-08-22 09:00:00',
    periods=10,
    freq='30min'
)
rng
```

```python
london = rng.tz_localize(tz="Europe/London")
london
```

```python
Calcutta = rng.tz_localize(tz="Asia/Calcutta")
Calcutta
```

```python
rng = pd.date_range(start="2026-08-22 09:00:00", periods=10, freq='30min')
s = pd.Series(range(10),index=rng)
l = s.tz_localize(tz='Europe/Berlin')
c = s.tz_localize(tz='Asia/Calcutta')
l + c
```

```python
dt = pd.date_range(start="2026-01-01 07:00:00",periods=11,freq='h')
d = pd.Series(range(11),index=dt)
l = d.tz_localize(tz='Europe/London')
c = l.tz_convert(tz='Asia/Calcutta')
l+c
```
