## Row Selection
- df.iloc[n] → position se row (0-indexed)
- df.loc[n] → label se row
- df.iloc[a:b] → end EXCLUDE
- df.loc[a:b] → end INCLUDE

```python
df.iloc[3]
```

```python
df.iloc[0:3]
```

```python
df.loc[3]
```

```python
df.loc[0:3]
```

```python
df.iloc[0:5]["Temperature"]
```

```python
df.loc[0:5, "Temperature"]
```
