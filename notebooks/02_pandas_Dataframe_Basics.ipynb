Statistical Operations:

df['col'].max() / .min() / .mean() / .std()
df.describe() → Complete summary (count, mean, std, min, 25%, 50%, 75%, max)
25% / 50% / 75% = Quartiles, meaning that percentage of data lies below that value

```python
df['Temperature'].max()
```

```python
df['Temperature'].min()
```

```python
df['Temperature'].mean()
```

```python
df['Temperature'].std()
```

Conditional Selection (Filtering):

df[df.Temperature >= 30] → Rows that satisfy the condition
df[df.Temperature == df['Temperature'].max()] → Row with the maximum value (DataFrame)
df['EST'][df.Temperature == df['Temperature'].max()] → Only a single column (Series)

```python
df[df.Temperature >= 30]
```

```python
df[df.Temperature == df['Temperature'].max()]
```

```python
df['EST'][df.Temperature == df['Temperature'].max()]
```
