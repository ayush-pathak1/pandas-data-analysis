pd.DataFrame(list_of_tuples, columns=[...]) → column names manual dene padte hain

```python
data = [
    ('Ayush', 22, 'Rampur'),
    ('Pari', 20, 'Satna'),
    ('King', 24, 'Unknown')
]

df = pd.DataFrame(data, columns= ['Name','Age','City'])
df
```
