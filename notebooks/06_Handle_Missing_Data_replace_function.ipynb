## Single Value Replace
- Kai baar datasets mein missing value ko special placeholder number se represent karte hain (jaise -99999)
- df.replace(old_value, new_value) se ise NaN ya kisi bhi sahi value se badal sakte hain

```python
import pandas as pd
import numpy as np
df = pd.read_csv('datasets/nyc_weather_data.csv')
new_df = df.replace(-99999,np.nan)
new_df
```

## List Se Replace — Multiple Placeholder Values Ek Saath
- Agar dataset mein missing data ko represent karne ke liye multiple special values hain (-99999, -88888), sabko ek saath NaN kar sakte hain

```python
new_df = df.replace([-99999,-88888],np.nan)
new_df
```

## Dictionary — Column-Specific Replace
- Har column ka apna "invalid value" define kiya ja sakta hai
- Real-world use: temperature/windspeed mein -99999 invalid hai, lekin event column mein '0' invalid hai (text column hai)

```python
import pandas as pd
import numpy as np
df = pd.read_csv('datasets/nyc_weather_data.csv')
new_df = df.replace({
    'temperature' : -99999,
    'windspeed' : -99999,
    'event' : '0'
}, np.nan)
new_df
```

## Dictionary — Value-to-Value Mapping (Bina Column Specify Kiye)
- Ye pattern column name nahi, balki "purani value: nayi value" batata hai
- Poore DataFrame mein jahan bhi ye value mile, replace ho jayegi

```python
new_df = df.replace({
    -99999 : np.nan,
    -88888 : np.nan,
    '0' : 'Sunny'
})
new_df
```

## Regex (Regular Expression) Se Replace
- regex=True dena zaroori hai, tabhi Pandas pattern ko literal text ki jagah "pattern rule" ki tarah samjhega
- [A-Za-z] → koi bhi English letter (capital ya small) match karega
- IMPORTANT MISTAKE TO AVOID: [A-za-z] (capital A se lowercase z) GALAT hai
  - ASCII table mein capital Z aur lowercase a ke beech special characters (`[`, `\`, `]`, `^`, `_`) bhi aate hain jo galti se match ho sakte hain
  - Hamesha [A-Za-z] hi likhna (capital-capital, phir small-small)

```python
new_df = df.replace({
    'temperature' : '[A-Za-z]',
    'windspeed' : '[A-Za-z]',
},'',regex=True)
new_df
```

## List to List — Categorical Data Ko Numeric Mein Convert Karna
- Position-wise mapping hoti hai: pehli list ka pehla item, dusri list ke pehle item se replace hota hai
- Real-world/ML use: categorical ratings (Poor/Average/Good) ko numeric score mein convert karna

```python
pd.set_option('future.no_silent_downcasting', True)
df = pd.DataFrame({
    'Score' : ['Good','Poor','Average','Exceptional','Good'],
   'Student' : ['Rohny','Monkesh','Rajni','Ayush','Pari']
})
new_df = df.replace(['Exceptional','Good','Average','Poor'],[90,70,60,40])
new_df
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Amit", "Rahul", "Priya", "Neha"],
    "Marks": [90, -99999, 80, -88888],
    "Grade": ["A", "0", "B", "0"]
})
new_df = df.replace({
    -99999 : np.nan,
    -88888 : np.nan,
    '0' : np.nan
})
new_df = new_df.fillna(50)
new_df
```
