## Basic concat() — Rows Jodना (Upar-Neeche)
- pd.concat([df1, df2]) → do DataFrames ko stack karta hai (rows add hoti hain)
- Har original DataFrame ka apna index (0,1,2...) rehta hai by default — isse duplicate index numbers ban sakte hain
- ignore_index=True → naya sequential index banata hai (0,1,2,3,4,5...), purane index clash nahi karte

```python
import pandas as pd

india_weather = pd.DataFrame({
    'city': ["Delhi", "mumbai", "Banglore"],
    'temperature': [32, 28, 45],
    'humidity': [60, 80, 78]
})

india_weather

usa_weather = pd.DataFrame({
    'city': ["new york", "chicago", "londob"],
    'temperature': [21, 12, 29],
    'humidity': [68, 65, 78]
})

usa_weather

df = pd.concat([india_weather,usa_weather], ignore_index=True)
df
```

## keys Parameter — Origin Track Karna
- keys=['label1', 'label2'] → batata hai kaunsi rows kis original DataFrame se aayi
- Result par df.loc['label'] laga ke sirf us source ka data nikaal sakte hain
- Useful jab multiple sources ka data combine karna ho, but unka origin yaad rakhna ho

```python
df = pd.concat([india_weather, usa_weather], keys=['india', 'usa'])
df
```

```python
df.loc['india']
```

## axis=1 — Columns Jodना (Side-by-Side)
- Default axis=0 (rows jodta hai, upar-neeche)
- axis=1 → DataFrames ko LEFT-RIGHT jodta hai (columns add hote hain, row-matching ke saath)
- Real use: agar alag-alag files mein alag-alag columns ka data ho (jaise ek file mein temperature, dusri mein windspeed), same cities ke liye, unhe side-by-side jodna

```python
temperature_df = pd.DataFrame({
    'city' : ['mumbai','delhi','jaipur'],
    'temprature' : [32,40,43]
})
temperature_df

windspeed_df = pd.DataFrame({
    'city' : ['mumbai','delhi','jaipur'],
    'windspeed' : [7,12,9]
})
windspeed_df

df = pd.concat([temperature_df,windspeed_df], axis=1)
df
```

## Series Ko DataFrame Mein Add Karna (concat Se)
- pd.Series([...], name='col_name') → ek Series banao, name= column ka naam banega
- axis=1 se concat karke ye Series DataFrame mein ek naye column ki tarah add ho jaati hai

```python
s = pd.Series(['Humid','Dry','Rain'], name='event')
df = pd.concat([temperature_df,s], axis=1)
df
```
