## Basic merge() = Inner Join (Default)
- pd.merge(df1, df2, on='column') → sirf wahi rows aayengi jo DONO DataFrames mein match karti hain
- Order of rows/DataFrames mein matter nahi karta, sirf matching VALUES pe based hota hai
- Default behavior "inner join" hi hota hai (how='inner' likhne ki zarurat nahi)

```python
import pandas as pd
df1 = pd.DataFrame({
    'city' : ['new york','chicago','orlando'],
    'temperature' : [21,14,25]
})
df1
```

```python
df2 = pd.DataFrame({
    'city' : ['chicago','new york','orlando'],
    'humidity' : [56,38,57]
})
df2
```

```python
df3 = pd.merge(df1,df2,on='city')
df3
```

## Outer Join
- how='outer' → SAARI rows aayengi (dono taraf se), match ho ya na ho
- Jaha match nahi milega, wahan NaN aa jaayega

```python
df4 = pd.DataFrame({
    'city' : ['new york','san francisco','chicago','orlando'],
    'temperature' : [21,14,25,28]
})
df4
```

```python
df5 = pd.DataFrame({
    'city' : ['ireland','chicago','new york','orlando'],
    'humidity' : [56,38,57,65]
})
df5
```

```python
df6 = pd.merge(df4,df5,on='city',how='outer')
df6
```

## indicator=True — Row Ka Origin Batana
- Ek naya column "_merge" add hota hai
- Values: "left_only", "right_only", ya "both"

```python
df6 = pd.merge(df4,df5,on='city',how='outer',indicator=True)
df6
```

## Left Join aur Right Join
- how='left' → LEFT DataFrame ki saari rows (jo sirf right mein hai, wo exclude ho jaayegi)
- how='right' → RIGHT DataFrame ki saari rows (jo sirf left mein hai, wo exclude ho jaayegi)
- Jo match nahi hota us taraf, NaN aa jaata hai

```python
df7 = pd.merge(df4,df5,on='city',how='left')
df7
```

```python
df8 = pd.merge(df4,df5,on='city',how='right')
df8 
```

## suffixes — Same-Naam Columns Ka Conflict Solve Karna
- Agar dono DataFrames mein same column naam hai (jaise dono mein temperature/humidity), Pandas confuse ho jaata
- Default: automatically "_x" (left) aur "_y" (right) laga deta hai
- suffixes=('_custom1', '_custom2') → apna naam de sakte ho

```python
df9 = pd.DataFrame({
    'city' : ['new york','chicago','orlando'],
    'temperature' : [21,14,25],
    'humidity' : [56,38,55]
})
df9
```

```python
df10 = pd.DataFrame({
    'city' : ['chicago','new york','orlando'],
    'humidity' : [38,56,65],
    'temperature' : [11,21,25],
})
df10
```

```python
df11 = pd.merge(df9,df10,on='city')
df11
```

```python
df12 = pd.merge(df9,df10,on='city', suffixes=('_left','_right'))
df12
```
