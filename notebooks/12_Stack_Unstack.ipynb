## Creating Multi-Level Columns (`header=[0,1]`)

* `pd.read_excel(path, header=[0,1], index_col=0)` → Parses Excel files where column headers span across two distinct rows
* Generates a **MultiIndex column** structure (e.g., top level: `'TCS'`, `'INFY'`; second level: `'Open'`, `'Close'`)

```python
import pandas as pd
df = pd.read_excel("../datasets/stocks.xlsx",header=[0,1], index_col=0)
df
```

## `stack()` — Rotating Innermost Column Headers into Row Index

* Pivots the innermost (lowest level) column header into the row index
* Reshapes the DataFrame into a tall/long format (similar in outcome to `melt()`)
* Operates on standard single-level DataFrames as well, treating the single column header row as the innermost level and converting it to row index entries
* `future_stack=True` → Enables the modern, recommended missing-data handling behavior and avoids deprecation warnings from legacy Pandas versions

```python
df.stack(future_stack=True)
```

```python
df_stacked = df.stack(future_stack=True)
df_stacked
```

## `unstack()` — The Inverse Operation of `stack()`

* Functions as the exact opposite of `stack()` — pivots the innermost (lowest level) row index back into column headers
* Reconstructs the original wide-format structure, completing the round-trip transformation

```python
df_stacked.unstack()
```

## `level` Parameter — Targeting Specific Levels in MultiIndex (3+ Levels)

* When columns contain 3 or more levels (e.g., Level 0: `Country`, Level 1: `Company`, Level 2: `Metric`), the `level` parameter explicitly specifies which index layer to stack or unstack
* Uses zero-based indexing (`level=0`, `level=1`, `level=2`) or exact level names passed as strings (e.g., `level='Metric'`)
* `level=2` → Targets and stacks the 3rd level (innermost level by default if unspecified)
* **Key Advantage:** Offers precise fine-grained control to restructure nested data without disturbing the outer hierarchical layers

```python
df1 = pd.read_excel("../datasets/stocks_3_levels.xlsx", header=[0,1,2], index_col=0)
df1
```

```python
df1.stack(level=2, future_stack=True)
```

## Summary Concept Connection — Simple vs. MultiIndex Reshaping

* **`melt()` / `pivot()**` → Primary tools for reshaping DataFrames with **flat (single-level)** column headers
* **`stack()` / `unstack()**` → Primary tools for reshaping DataFrames with **hierarchical (MultiIndex)** row or column headers
* **Core Concept:** Both pairs share the exact same underlying purpose: seamlessly transitioning data between **wide format** and **long format**
