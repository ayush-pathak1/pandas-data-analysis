# Lecture 12: stack() aur unstack() — MultiIndex Reshaping
## Multi-Level Columns Banana (header=[0,1])
- pd.read_excel(path, header=[0,1], index_col=0) → Excel file mein agar column headers 2 rows mein faile hain
- Isse MultiIndex columns bante hain (jaise "TCS"→"Open"/"Close", "INFY"→"Open"/"Close")

```python
import pandas as pd
df = pd.read_excel("datasets/stocks.xlsx",header=[0,1], index_col=0)
df
```

## stack() — Innermost Column Level Ko Rows Mein Convert Karna
- Columns ke sabse andar (innermost) level ko row/index ka part bana deta hai
- Result "long/tall" format ban jaata hai (jaisa melt() mein hota hai)
- Simple (non-MultiIndex) DataFrame pe bhi kaam karta hai — us case mein normal columns hi "innermost level" maane jaate hain aur wahi rows mein convert hote hain
- future_stack=True → naye/recommended NaN-handling behavior ke liye (deprecated purane behavior se bachne ke liye)

```python
df.stack(future_stack=True)
```

```python
df_stacked = df.stack(future_stack=True)
df_stacked
```

## unstack() — Reverse Operation
- stack() ka opposite — rows ke innermost level ko wapas columns mein badal deta hai
- Original wide format wapas mil jaata hai

```python
df_stacked.unstack()
```

## level Parameter — 3+ Levels Ke Liye
- Jab columns mein 3 ya zyada levels hon (jaise Country → Company → Metric)
- level=n (0-indexed) se batate hain KONSA level stack karna hai
- level=2 → teesra (sabse andar wala) level

```python
df1 = pd.read_excel("datasets/stocks_3_levels.xlsx", header=[0,1,2], index_col=0)
df1
```

```python
df1.stack(level=2, future_stack=True)
```

## Concept Connection (Yaad Rakhna)
- melt()/pivot() → simple flat columns ke liye reshaping
- stack()/unstack() → MultiIndex (multi-level) columns/index ke liye reshaping
- Dono ka core idea same hai: wide format ↔ long format ke beech convert karna
