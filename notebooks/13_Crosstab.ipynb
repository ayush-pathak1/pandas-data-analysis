```python
import pandas as pd
df = pd.read_excel("../datasets/survey.xls")
df
```

Purpose: Creates a basic frequency table showing counts of people grouped by Nationality (rows) and Handedness (columns: Left vs. Right handed).

```python
pd.crosstab(df.Nationality,df.Handedness)
```

Purpose: Displays the distribution count between Sex (rows) and Handedness (columns).

```python
pd.crosstab(df.Sex,df.Handedness)
```

Purpose: Generates the same table as Cell 3, but margins=True adds an "All" row and column displaying the total counts for each group and the overall grand total.

```python
pd.crosstab(df.Sex,df.Handedness, margins=True)
```

Purpose: Groups Sex on rows and creates nested columns for Handedness and Nationality combined, along with grand totals.

```python
pd.crosstab(df.Sex, [df.Handedness,df.Nationality], margins=True)
```

Purpose: Creates nested rows combining Nationality and Sex, against Handedness on columns, showing structured sub-totals and grand totals (margins=True).

```python
pd.crosstab([df.Nationality, df.Sex], [df.Handedness], margins=True)
```

Purpose: Displays proportions instead of raw counts. normalize='index' calculates percentages row-wise, showing the fraction of Left/Right-handed individuals within each Sex (each row adds up to 1.0 or 100%).

```python
pd.crosstab(df.Sex, df.Handedness, normalize='index')
```

Purpose: Calculates the average Age for each combination of Sex and Handedness (e.g., average age of Left-handed Females). values=df.Age defines the target column, and aggfunc=np.average sets the calculation rule.

```python
import numpy as np
pd.crosstab(df.Sex, df.Handedness, values=df.Age, aggfunc=np.average)
```
