# Lecture 18: Handling Holidays in Time Series

## CustomBusinessDay + USFederalHolidayCalendar
- freq='B' skips only weekends
- USFederalHolidayCalendar skips weekends + US federal holidays
- CustomBusinessDay(calendar=USFederalHolidayCalendar()) creates a custom frequency

```python
import pandas as pd
from pandas.tseries.holiday import USFederalHolidayCalendar
from pandas.tseries.offsets import CustomBusinessDay

df = pd.read_csv("../datasets/aapl_no_dates.csv")

# Normal business days vs US Calendar
pd.date_range(start="7/1/2017", end="7/21/2017", freq='B')
```

```python
usb = CustomBusinessDay(calendar=USFederalHolidayCalendar())
rng = pd.date_range(start="7/01/2017", end="7/21/2017", freq=usb)
rng
```

```python
df.set_index(rng, inplace=True)
```

## Custom Holiday Calendar
- Inherit from AbstractHolidayCalendar
- Define rules = [Holiday('name', month=, day=)] 
- observance=nearest_workday shifts holiday to nearest workday if it falls on weekend

```python
# Custom calendar
from pandas.tseries.holiday import AbstractHolidayCalendar, nearest_workday, Holiday

class myCalendar(AbstractHolidayCalendar):
    rules = [
        Holiday('My Birth Day', month=11, day=27),
    ]

my_bday = CustomBusinessDay(calendar=myCalendar())
pd.date_range('11/1/2017', '11/30/2017', freq=my_bday)
```

## weekmask
- CustomBusinessDay(weekmask='Mon Tue Wed Thu Sun') → custom working days
- holidays=[...] → extra holidays to skip

```python
# weekmask
b = CustomBusinessDay(weekmask='Mon Tue Wed Thu Sun')
pd.date_range(start="7/1/2017", end="7/29/2017", freq=b)
```

```python
b = CustomBusinessDay(weekmask='Mon Tue Wed Thu Sun', holidays=["2017-07-04"])
pd.date_range(start="7/1/2017", end="7/29/2017", freq=b)
```
