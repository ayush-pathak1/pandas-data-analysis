# Lecture 19: to_datetime() — Converting Strings to Datetime

## Basic conversion
- pd.to_datetime(dates, format='mixed') → handles multiple formats in one list
- dayfirst=True → treats first number as day (European format: DD/MM/YYYY)
- format='%d$%m$%Y' → custom separator/format string

```python
import pandas as pd

dates = ['2017-01-05', 'jan 5, 2017', '01/05/2017', '2017.01.05', '20170105']
pd.to_datetime(dates, format='mixed')
```

```python
dts = ['2017-01-05 2:30:00 PM', 'jan 5, 2017 14:30:00', '01/05/2017', '2017.01.05', '20170105']
pd.to_datetime(dts, format='mixed')
```

## Common format codes
- %d = day, %m = month, %Y = 4-digit year, %H = hour, %M = minute, %S = second

```python
pd.to_datetime('5/1/2017')
```

```python
pd.to_datetime('5/1/2017', dayfirst=True)
```

```python
pd.to_datetime('5$1$2017', format='%d$%m$%Y')
```

## Epoch/Unix timestamp
- pd.to_datetime(t, unit='s') → converts seconds since 1970-01-01 to datetime
- dt.view('int64') → converts datetime back to integer timestamp

```python
t = 1789936167
pd.to_datetime(t, unit='s')
```

```python
dt = pd.to_datetime([t], unit='s')
dt.view('int64')
```

## Error handling
- errors='coerce' → invalid strings become NaT (not an error)
- errors='raise' → default, throws error on invalid string
- errors='ignore' → returns original string unchanged

```python
pd.to_datetime(['2023-01-15', 'not a date', '2023-03-20'], errors='coerce')
```
