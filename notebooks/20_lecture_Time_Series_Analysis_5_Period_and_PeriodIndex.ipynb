 Lecture 20: Period and PeriodIndex

## Timestamp vs Period
- Timestamp → a specific point in time (e.g. "2026-01-05 14:30")
- Period → a duration/span of time (e.g. "January 2026", "Q1 2026")

```python
import pandas as pd 
y  = pd.Period("2016")
y
```

```python
y.start_time
```

```python
y.end_time
```

## pd.Period()
- pd.Period("2016") → yearly period
- pd.Period("2026-1", freq='M') → monthly period
- pd.Period("2026-11-27", freq='D') → daily period
- pd.Period("2026Q1") → quarterly period
- .start_time → start of the period
- .end_time → end of the period
- period + n → moves n periods forward (e.g. m+1 = next month)

```python
m = pd.Period("2026-1",freq='M')
m
```

```python
m.start_time
```

```python
m.end_time
```

```python
m+1
```

```python
d = pd.Period("2026-11-27", freq='D')
d
```

```python
d+1
```

```python
h = pd.Period("2026-11-27 23:00:00", freq='H')
h
```

```python
q = pd.Period("2026Q1")
q
```

```python
q+1
```

## Quarterly Period with Custom Fiscal Year
- pd.Period("2026Q1", freq='Q-JAN') → Q1 ends in January
- .asfreq('M', how='end') → last month of that quarter
- .asfreq('M', how='start') → first month of that quarter
- q2 - q1 → difference in periods (returns integer)

```python
new_q = pd.Period("2026Q1",freq='Q-JAN')
new_q
```

```python
new_q.start_time
```

```python
new_q.asfreq('M',how="end")
```

```python
new_q.asfreq('M',how="start")
```

```python
q = pd.Period("2026Q1",freq='Q-JAN')
```

```python
q2 = pd.Period("2026Q2", freq='Q-JAN')
q2
```

```python
q2-q
```

## period_range()
- pd.period_range(start, end, freq=) → range between two dates
- pd.period_range(start, periods=n, freq=) → n periods from start
- freq='Q-JAN' → quarterly, fiscal year ends January

```python
idx = pd.period_range('2019', '2026', freq='Q')
idx
```

```python
idx = pd.period_range('2019','2026', freq='Q-JAN')
idx
```

```python
idx  = pd.period_range('2019',periods=10,freq='Q-JAN')
idx
```

## PeriodIndex with Series
- ps = pd.Series(data, index=period_range) → Series with PeriodIndex
- ps['2021'] → partial string indexing (all of 2021)
- ps['2019':'2021'] → range slicing

```python
import numpy as np
ps = pd.Series(np.random.randn(len(idx)), idx)
ps
```

```python
ps['2019']
```

```python
ps['2019':'2021']
```

# Converting Between PeriodIndex and DatetimeIndex
- ps.to_timestamp() → PeriodIndex → DatetimeIndex
- pst.to_period() → DatetimeIndex → PeriodIndex
- idx[0].start_time / idx[0].end_time → access start/end of any period

```python
pst = ps.to_timestamp()
pst
```

```python
pst.to_period()
```

# PeriodIndex on a Real DataFrame
- df1.T → transpose (rows become columns, columns become rows)
- pd.PeriodIndex(df1.index, freq='Q-JAN') → converts string index to PeriodIndex
- df1.index.map(lambda x: x.start_time) → new column with period start dates
- df1.index.map(lambda x: x.end_time) → new column with period end dates

```python
import pandas as pd
df1 = pd.read_csv("../datasets/wmt.csv")
df1
```

```python
df1.set_index('Line Item',inplace=True)
df1 = df1.T
```

```python
df1.index = pd.PeriodIndex(df1.index, freq='Q-JAN')
df1.index
```

```python
df1['start date'] = df1.index.map(lambda x: x.start_time)
df1
```

```python
df1['end date'] = df1.index.map(lambda x: x.start_time)
df1
```
