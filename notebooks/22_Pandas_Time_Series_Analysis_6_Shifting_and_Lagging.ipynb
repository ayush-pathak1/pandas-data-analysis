# Lecture 22: shift() — Moving Data Forward or Backward

## shift(n)
- df.shift(1) → moves all values DOWN by 1 row (first row becomes NaN)
- df.shift(-1) → moves all values UP by 1 row (last row becomes NaN)
- Useful for comparing current value with previous/next value

```python
import pandas as pd
df = pd.read_csv("../datasets/fb.csv",parse_dates=['Date'],index_col='Date')
df
```

```python
df.shift(2)
```

```python
df.shift(-1)
```

## Common Real-World Uses
- Previous day price: df['Price'].shift(1)
- Daily price change: df['Price'] - df['Price'].shift(1)
- 5-day return %: (df['Price'] - df['Price'].shift(5)) * 100 / df['Price'].shift(5)

```python
df['prev day price'] = df['Price'].shift(1)
df
```

```python
df['1 Day change'] = df['Price'] - df['prev day price']
df
```

```python
df['5 day % return'] = (df['Price'] - df['Price'].shift(5))*100/df['Price'].shift(5)
df
```

## tshift(n) — Shift the INDEX (not the data)
- Shifts the DatetimeIndex forward/backward by n periods
- Data stays in same rows, only timestamps change
- shift() moves DATA, df.shift(1, freq="B") moves TIMESTAMPS

```python
df = df[['Price']]
df
```

```python
df_shifted = df.shift(1, freq="B")
df_shifted
```
