### header=None aur names= 
- header=None → file mein pehli row bhi data hai, header nahi
- names=[...] → manually column names dena

```python
df = pd.read_csv("datasets/stock_data.csv",header=None)
df
```

```python
df = pd.read_csv("datasets/stock_data.csv", names = ["ticker", "eps","revenue","price", "people"])
df
```

### nrows — Limited rows padhna (bade files test karne ke liye)

```python
df = pd.read_csv("datasets/stock_data.csv", nrows=2)
df
```

### na_values — Messy data ko NaN banana
- List wala → sabhi columns pe same rule
- Dictionary wala → har column ka apna alag rule (column-specific)

```python
df = pd.read_csv("datasets/stock_data.csv", na_values=["not available", "n.a."])
df
```

```python
df = pd.read_csv("datasets/stock_data.csv",na_values={
    'eps' : ["not available", "n.a."],
    'revenue' : ["not available", "n.a.", -1],
    'people' : [" not available", "n.a."]
})
df
```

- header=False → column names save nahi honge

```python
df.to_csv("new.csv")
df
```

```python
df.to_csv("new.csv",index=False)
df
```

### read_excel() aur converters
- converters → kisi column ki har value ko custom function se transform karna

```python
df = pd.read_excel("datasets/stock_data.xlsx", sheet_name= "Sheet1")
df
```

```python
def convert_people_cell(cell):
    if cell == "n.a." :
        return "sam walton"
    return cell
df = pd.read_excel("datasets/stock_data.xlsx", sheet_name="Sheet1" , converters={'people':convert_people_cell})
df
```

### to_excel() — startrow/startcol se position control

```python
df.to_excel("new.xlsx",sheet_name= 'stocks')
df
```

```python
df.to_excel("new.xlsx",sheet_name= "stocks", startrow=1, startcol=2)
```

### ExcelWriter — Ek file mein multiple sheets (multiple DataFrames)

```python
df_stocks = pd.DataFrame({
    'TICKER' : ['GOOGLE', 'APPLE'],
    'EPS' : [34,23],
    'ROE' : [54, 63]
})

df_weather = pd.DataFrame({
    'DATE' : ['1/1/2017','1/2/2017','1/3/2017'],
    'TEMPRATURE' : [32,30,28],
    'EVENTS' : ['SUNNY', 'CLOUDY', 'RAIN']
})

with pd.ExcelWriter('Stock_Weather.xlsx') as writer:
    df_stocks.to_excel(writer, sheet_name="stocks")
    df_weather.to_excel(writer, sheet_name="weather")
```
