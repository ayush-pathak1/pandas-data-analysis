header=None and names=

header=None → The first row in the file is actual data, not a header
names=[...] → Manually provide column names

```python
df = pd.read_csv("../datasets/stock_data.csv",header=None)
df
```

```python
df = pd.read_csv("../datasets/stock_data.csv", names = ["ticker", "eps","revenue","price", "people"])
df
```

nrows — Reading a limited number of rows (for testing large files)

```python
df = pd.read_csv("../datasets/stock_data.csv", nrows=2)
df
```

na_values — Converting messy data into NaN :

Using a List → Applies the same rule across all columns
Using a Dictionary → Applies column-specific rules to each column

```python
df = pd.read_csv("../datasets/stock_data.csv", na_values=["not available", "n.a."])
df
```

```python
df = pd.read_csv("../datasets/stock_data.csv",na_values={
    'eps' : ["not available", "n.a."],
    'revenue' : ["not available", "n.a.", -1],
    'people' : [" not available", "n.a."]
})
df
```

to_csv() — Saving a DataFrame to a CSV file :

index=False → Excludes the row index from being saved (best practice)
columns=[...] → Saves only specific columns
header=False → Excludes column names from being saved

```python
df.to_csv("new.csv")
df
```

```python
df.to_csv("new.csv",index=False)
df
```

```python
df.columns
```

```python
df.to_csv("new.csv",columns=['ticker', 'eps'])
df
```

```python
df.to_csv("new.csv", header=False)
df
```

read_excel() and converters
converters → Transforms every value in a specific column using a custom function

```python
df = pd.read_excel("../datasets/stock_data.xlsx", sheet_name= "Sheet1")
df
```

```python
def convert_people_cell(cell):
    if cell == "n.a." :
        return "sam walton"
    return cell
df = pd.read_excel("../datasets/stock_data.xlsx", sheet_name="Sheet1" , converters={'people':convert_people_cell})
df
```

to_excel() — Controlling position using startrow / startcol

nrows — Reading a limited number of rows (for testing large files)

```python
df.to_excel("new.xlsx",sheet_name= 'stocks')
df
```

```python
df.to_excel("new.xlsx",sheet_name= "stocks", startrow=1, startcol=2)
```

ExcelWriter — Saving multiple sheets in a single file (multiple DataFrames)

```python
df_stocks = pd.DataFrame({
    'TICKER' : ['GOOGLE', 'APPLE'],
    'EPS' : [34,23],
    'ROE' : [54, 63]
})

df_weather = pd.DataFrame({
    'DATE' : ['1/1/2017','1/2/2017','1/3/2017'],
    'TEMPRATURE' : [32,30,28],
    'EVENTS' : ['SUNNY', 'CLOUDY', 'RAIN']
})

with pd.ExcelWriter('Stock_Weather.xlsx') as writer:
    df_stocks.to_excel(writer, sheet_name="stocks")
    df_weather.to_excel(writer, sheet_name="weather")
```
